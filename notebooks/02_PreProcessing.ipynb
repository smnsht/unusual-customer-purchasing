```python
import pandas as pd
```

## Load all transactions

Load the retail transactions dataset into a pandas DataFrame.

```python
df = pd.read_csv("../data/online_retail_II.csv")
df.head()
```

## Drop non-UK transactions

Keep only rows where `Country` is `United Kingdom`.

```python
df.drop(df[df["Country"] != "United Kingdom"].index, inplace=True)
```

## Remove rows with missing `Customer ID`

Exclude guest customers because they are not needed for this analysis.

```python
df.dropna(subset=["Customer ID"], inplace=True)
```

## Drop unused columns

Remove `Country` and `Description` because they are not required for the next steps.

```python
df.drop(columns=["Description", "Country"], inplace=True)
```

## Convert `InvoiceDate` to a datetime value

Parse the `InvoiceDate` column so it can be used in time-based analysis.

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

## Inspect data types

Review the DataFrame schema to confirm that each column has the expected type.

```python
df.info()
```

## Check for missing values

Verify that no unexpected missing values remain in the cleaned dataset.

```python
df.isnull().sum()
```

## Export the cleaned dataset

Save the cleaned UK transactions dataset as a CSV file for later use.

```python
df.to_csv("../data/transactions_uk_cleaned.csv", index=False)
```
